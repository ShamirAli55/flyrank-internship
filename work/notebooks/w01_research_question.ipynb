## 1. My lane (or freestyle) and why

I picked **Content Refresh Prioritization**. The basic problem is that a website with thousands of pages can't review everything at once — you need some way to decide which pages to look at first. I wanted to see if the search performance data already in the dataset (impressions, position, CTR, content age) is enough to make that call more consistently than just eyeballing it.

It also felt like a concrete problem where a ranked output actually connects to something a real person would act on.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Total pages:", len(df))
print("\nTrend Direction Distribution:")
print(df["trend_direction"].value_counts())

print("\nAverage CTR:", round(df["ctr"].mean(), 4))
print("Average Content Age (days):", round(df["content_age_days"].mean(), 2))
```

## 2. The question: decision, action, cost of a wrong call

**Decision:** Which pages should be flagged for a content review?

**Who acts on it:** A content manager. They get a ranked list and decide page by page whether to update. The model doesn't update anything it just helps with triage.

**Cost of a wrong call:** Two kinds:
- Miss a declining page → it keeps sliding without anyone noticing. That's the more costly error because it's invisible.
- Flag a page that's fine → wastes a reviewer's time on something that doesn't need touching.

Neither error is free, but the first one compounds over time, so the ranking should lean toward catching actual decliners near the top.

```python
# Declining pages are over half the dataset — not a rare edge case
declining = (df["trend_direction"] == "down").sum()
total = len(df)
print(f"Declining pages: {declining} ({declining/total*100:.1f}%)")
print(f"Non-declining pages: {total - declining} ({(total-declining)/total*100:.1f}%)")
```

## 3. Quick look at the data (2-3 real numbers)

- **16,262 pages** are already trending down — that's 54% of the dataset. There's clearly a real problem to prioritize.
- **Average content age: 256 days** — most of these pages haven't been touched in almost a year.
- **Average CTR: 0.51%** — low overall, which means there's headroom to improve if the right pages get refreshed.

## 4. Careful words: what I can and can't claim

**Can claim:** associations between measured signals (impressions, CTR, position, content age) and the `trend_direction` label. Any finding is observed and directional — it's a pattern in this dataset, not a formula for rankings.

**Cannot claim:** why a page is declining, whether an update will fix it, or anything about Google's algorithm. The label itself (`trend_direction == 'down'`) is derived from the data, not an external ground truth.

The output is decision-support: it helps a human decide where to look, not a guarantee of what they'll find.

```python
# These columns encode the outcome — cannot be used as model features
leaky = [c for c in df.columns if "trend" in c]
print("Columns that must stay out of features (they reveal the label):")
print(leaky)
```
